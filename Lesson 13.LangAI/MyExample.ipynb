# Задача 1: Класифікація текстів

## 🧾 Як представити текст: від слів до чисел

Машини не працюють безпосередньо з текстом — їм потрібні **числові вектори**.  
Тому перший крок у NLP — **перетворення тексту в числове представлення**. Існує кілька підходів:

---

### ✂️ 1. Токенізація

**Токенізація** — це розбиття тексту на окремі елементи: слова, речення або підслова.

**Приклад:**
“Я люблю машинне навчання” → [“Я”, “люблю”, “машинне”, “навчання”]

---

### 🧮 2. Bag of Words (BoW)

Модель BoW представляє текст як **частоти слів**, ігноруючи порядок.

**Приклад:**

| Слово       | Частота |
|-------------|---------|
| машинне     | 1       |
| навчання    | 1       |
| люблю       | 1       |
| Я           | 1       |

Недоліки:

- Не враховує порядок слів
- Велика розмірність для великих словників

---

### 🔍 3. TF-IDF (Term Frequency — Inverse Document Frequency)

TF-IDF зменшує вагу "поширених" слів (як "і", "це", "я") і підсилює важливіші.

**Формула:**

$$TF-IDF = TF * log(N / DF)$$

де:
- **TF** — частота терміну в документі  
- **DF** — кількість документів з цим терміном  
- **N** — загальна кількість документів

TF-IDF — хороший компроміс між простотою BoW і складністю сучасних embeddings.


```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer  # Імпортуємо CountVectorizer для BoW і TfidfVectorizer для TF-IDF

# Приклад корпусу
texts = [  # Створюємо список текстових документів
    "Я люблю машинне навчання",  # Перший документ
    "Машинне навчання — це цікаво",  # Другий документ
    "Я вивчаю машинне навчання щодня"  # Третій документ
]

# --- Bag of Words ---
bow_vectorizer = CountVectorizer()  # Створюємо об'єкт для перетворення тексту в BoW
bow_matrix = bow_vectorizer.fit_transform(texts)  # Навчаємо векторизатор на текстах і перетворюємо їх у матрицю

print("🎒 Bag of Words (BoW):")  # Виводимо заголовок результату BoW
print(bow_vectorizer.get_feature_names_out())  # Виводимо всі слова, які стали ознаками
print(bow_matrix.toarray())  # Перетворюємо розріджену матрицю у звичайний масив і виводимо її

# --- TF-IDF ---
tfidf_vectorizer = TfidfVectorizer()  # Створюємо об'єкт для TF-IDF перетворення
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)  # Навчаємо векторизатор і перетворюємо тексти у TF-IDF матрицю

print("\n📊 TF-IDF:")  # Виводимо заголовок результату TF-IDF
print(tfidf_vectorizer.get_feature_names_out())  # Виводимо слова, для яких обчислено TF-IDF
print(tfidf_matrix.toarray())  # Перетворюємо розріджену матрицю у звичайний масив і виводимо її
```

```python
# ----------------- ЦЕЙ КОД НЕ ПРАЦЮЄ У МЕНЕ -----------------
# Змінив версію Python і почало працювати: Python 3.12

from sklearn.datasets import fetch_20newsgroups  # Імпортуємо функцію для завантаження датасету 20 Newsgroups
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer  # Імпортуємо методи BoW та TF-IDF
from sklearn.linear_model import LogisticRegression  # Імпортуємо логістичну регресію для класифікації тексту
from sklearn.model_selection import train_test_split  # Імпортуємо функцію для поділу даних на train і test
from sklearn.metrics import classification_report  # Імпортуємо звіт з precision, recall та F1-score

# 1. Завантажуємо текстовий датасет (2 категорії для простоти)
categories = ['sci.space', 'rec.sport.hockey']  # Вибираємо лише дві категорії: космос і хокей
data = fetch_20newsgroups(  # Завантажуємо датасет 20 Newsgroups
    subset='all',  # Використовуємо всі доступні дані: train + test
    categories=categories,  # Завантажуємо тільки вибрані дві категорії
    remove=('headers', 'footers', 'quotes')  # Видаляємо заголовки, підписи та цитати з текстів
)

# print("Hello my friends")  # Тестовий print, зараз закоментований і не виконується
data.data[0]  # Виводимо перший текст із датасету
```

```python
print(f"Кількість документів: {len(data.data)}")  # Виводимо загальну кількість текстових документів
print(f"Категорії: {data.target_names}")  # Виводимо назви категорій

print("\nПерший документ:\n")  # Виводимо заголовок перед першим документом
print(data.data[0][:1000])  # Виводимо перші 1000 символів першого документа
```

```python
from sklearn.linear_model import LogisticRegression  # Імпортуємо модель логістичної регресії для класифікації
from sklearn.model_selection import train_test_split  # Імпортуємо функцію для поділу даних на train і test
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer  # Імпортуємо BoW та TF-IDF векторизатори
from sklearn.metrics import classification_report  # Імпортуємо звіт для оцінки якості моделі

# 2. Ділимо на train/test
X_train, X_test, y_train, y_test = train_test_split(  # Ділимо тексти та їх класи на навчальну і тестову частини
    data.data,  # Передаємо всі текстові документи
    data.target,  # Передаємо правильні класи для кожного документа
    test_size=0.3,  # 30% даних використовуємо для тестування, 70% для навчання
    random_state=42  # Фіксуємо випадковість, щоб при повторному запуску отримати той самий поділ
)

# 3. Векторизація тексту (спробуй BoW або TF-IDF)
#vectorizer = CountVectorizer()  # Створюємо BoW-векторизатор, який рахує кількість появи слів
vectorizer = TfidfVectorizer()  # Створюємо TF-IDF-векторизатор, який оцінює важливість слів

X_train_vec = vectorizer.fit_transform(X_train)  # Навчаємо TF-IDF на train-текстах і перетворюємо їх у числові вектори
X_test_vec = vectorizer.transform(X_test)  # Перетворюємо test-тексти за словником, отриманим тільки з train

# 4. Тренування моделі
clf = LogisticRegression(max_iter=1000)  # Створюємо модель логістичної регресії з максимумом 1000 ітерацій
clf.fit(X_train_vec, y_train)  # Навчаємо модель на векторизованих текстах та правильних класах

# 5. Оцінка
y_pred = clf.predict(X_test_vec)  # Модель прогнозує класи для тестових текстів
print(classification_report(y_test, y_pred, target_names=data.target_names))  # Виводимо precision, recall, F1-score та accuracy
```

## ✂️ Стемінг і лемматизація

Перш ніж векторизувати текст, часто варто його **нормалізувати** — звести слова до базової форми.  
Це допомагає зменшити розмір словника та об'єднати варіанти одного слова.

---

### 🌱 Стемінг (Stemming)

**Стемінг** — це грубе обрізання слів до кореня (стему).  
Метод швидкий, але іноді "ламає" слова.

**Приклад:**
“навчаю”, “навчання”, “навчався” → “навч”

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
print(stemmer.stem("learning"))     # → 'learn'
print(stemmer.stem("learned"))      # → 'learn'
```

## 🌳 Лемматизація (Lemmatization)

Лемматизація — це зведення слова до його словникової форми (леми).
Метод точніший за стемінг, але повільніший і потребує знання частини мови.

Приклад:
"навчаю", "навчання", "навчався" → "навчати"

```python
from nltk.stem import WordNetLemmatizer  # Імпортуємо лематизатор, який приводить слова до початкової форми
import nltk  # Імпортуємо бібліотеку NLTK для роботи з природною мовою

nltk.download('wordnet')  # Завантажуємо словник WordNet, необхідний для лематизації

lemmatizer = WordNetLemmatizer()  # Створюємо об'єкт лематизатора

print(lemmatizer.lemmatize("learning", pos="v"))  # Лематизуємо "learning" як дієслово → "learn"
print(lemmatizer.lemmatize("better", pos="a"))  # Лематизуємо "better" як прикметник → "good"
```

```python
import nltk  # Імпортуємо бібліотеку NLTK для роботи з текстом
from nltk.stem import PorterStemmer  # Імпортуємо Porter Stemmer для стемінгу слів
from nltk.tokenize import word_tokenize  # Імпортуємо токенізатор для розбиття речення на слова

nltk.download('punkt_tab')  # Завантажуємо дані, необхідні для роботи word_tokenize

stemmer = PorterStemmer()  # Створюємо об'єкт стемера Porter

# Функція для стемінгу речення
def apply_stemming(sentence):  # Створюємо функцію, яка отримує одне речення
    words = word_tokenize(sentence)  # Розбиваємо речення на окремі слова
    stemmed_words = [stemmer.stem(word) for word in words]  # Для кожного словамо його основу
    return ' '.join(stemmed_words)  # Об'єднуємо оброблені слова отримує назад у речення

# Припустимо, data.data — це список текстових документів
stemmed_train_data = [apply_stemming(sentence) for sentence in data.data]  # Застосовуємо стемінг до кожного документа

from sklearn.feature_extraction.text import TfidfVectorizer  # Імпортуємо TF-IDF векторизатор

vectorizer = TfidfVectorizer()  # Створюємо TF-IDF векторизатор
converted = vectorizer.fit_transform(stemmed_train_data)  # Навчаємо TF-IDF на оброблених текстах і перетворюємо їх у числову матрицю
print(converted.shape)  # Виводимо розмір отриманої TF-IDF матриці
```

# 🧩 Задача 2: Класифікація токенів

**Класифікація токенів** — це тип задачі в NLP, де модель присвоює **кожному токену в тексті** певну категорію або мітку.

---

### 📌 Типові задачі класифікації токенів:

| Задача                      | Приклад міток           | Для чого це |
|----------------------------|--------------------------|-------------|
| 🔤 **POS-tagging**         | `NOUN`, `VERB`, `ADJ`   | Частина мови |
| 🏷️ **NER**                | `B-PER`, `I-ORG`, `O`   | Іменовані сутності |
| 🧱 **Chunking**            | `B-NP`, `I-NP`, `O`      | Іменні групи |
| 🪓 **Segmentation**        | `B`, `I`, `E`, `S`       | Розбиття тексту |
| 🔐 **Masked Language Modeling** | `MASK` або індекс мітки | Передбачення токенів |

---

### 🤖 Як це працює?

1. **Текст розбивається на токени**
2. **Кожному токену призначається мітка**
3. **Модель навчається передбачати мітки для нових токенів**

---

### 📐 Формат даних

Зазвичай у форматі **послідовності** токенів і відповідних **міток**:

```python
import nltk  # Імпортуємо бібліотеку NLTK для роботи з природною мовою
from nltk.corpus import conll2002  # Імпортуємо датасет CoNLL-2002

# cat dog mouse
# [1 0 0]
# [0 1 0]

nltk.download('conll2002')  # Завантажуємо датасет CoNLL-2002, якщо його ще немає

# Load the train and test sets
train_sents = conll2002.tagged_sents('ned.train')  # Завантажуємо навчальні речення з нідерландського набору
test_sents = conll2002.tagged_sents('ned.testa')  # Завантажуємо тестові речення з нідерландського набору

# print("train_sents", len(train_sents))
# print("train_sents[0]", train_sents[0])

# print("test_sents", len(test_sents))
# print("test_sents[0]", test_sents[0])

# Extract word-tag pairs
def extract_word_tag_pairs(sentences):  # Створюємо функцію для отримання пар слово-тег
    processed_sentences = []  # Створюємо порожній список для оброблених речень

    for sentence in sentences:  # Перебираємо всі речення
        processed_sentences.append([(word, tag) for word, tag in sentence])  # Додаємо кожне слово разом із його тегом

    return processed_sentences  # Повертаємо список оброблених речень

train_data = extract_word_tag_pairs(train_sents)  # Обробляємо навчальний набір
test_data = extract_word_tag_pairs(test_sents)  # Обробляємо тестовий набір

print(len(train_data), len(test_data))  # Виводимо кількість речень у train та test

# Print the first few sentences to inspect the data format
print("First 5 training sentences:")  # Виводимо заголовок для навчальних речень

for i in range(min(5, len(train_data))):  # Перебираємо перші 5 речень або менше, якщо їх недостатньо
    print(train_data[i])  # Виводимо поточне навчальне речення

print("\nFirst 5 testing sentences:")  # Виводимо заголовок для тестових речень

for i in range(min(5, len(test_data))):  # Перебираємо перші 5 тестових речень або менше
    print(test_data[i])  # Виводимо поточне тестове речення
```

```python
# Імпортуємо модель логістичної регресії.
# Вона буде використовуватися для класифікації слів за їхніми тегами.
from sklearn.linear_model import LogisticRegression

# Імпортуємо функцію для розділення даних на навчальну та тестову вибірки.
from sklearn.model_selection import train_test_split

# Імпортуємо функцію для побудови звіту про якість класифікації.
# Звіт міститиме precision, recall, f1-score та support.
from sklearn.metrics import classification_report

# Імпортуємо DictVectorizer.
# Він перетворює словники ознак у числову матрицю,
# яку може використовувати машинна модель.
from sklearn.feature_extraction import DictVectorizer


# Об'єднуємо навчальні та тестові речення в один список.
# train_data містить навчальні речення.
# test_data містить тестові речення.
# Після цього all_data містить усі речення разом.
all_data = train_data + test_data


# Створюємо порожній список для початкових ознак.
# У нього будемо додавати словники з ознаками кожного слова.
X_raw = []


# Створюємо порожній список для правильних відповідей (міток).
# Тут будуть зберігатися POS-теги слів, наприклад:
# NN, VB, JJ, PRP тощо.
y = []


# Проходимо по кожному реченню у всьому наборі даних.
for sentence in all_data:

    # Проходимо по кожному слову та його тегу в поточному реченні.
    # word — саме слово.
    # tag — його правильний граматичний тег.
    for word, tag in sentence:

        # Створюємо словник ознак для поточного слова.
        # Наприклад:
        # {"word": "cat"}
        #
        # Поки що використовуємо тільки одну ознаку — саме слово.
        X_raw.append({"word": word})

        # Додаємо правильний тег цього слова до списку y.
        # Наприклад:
        # y = ["NN", "VB", "DT", ...]
        y.append(tag)


# Створюємо об'єкт DictVectorizer.
# sparse=True означає, що результат буде зберігатися
# у вигляді розрідженої матриці (sparse matrix).
vec = DictVectorizer(sparse=True)


# Перетворюємо словники X_raw у числову матрицю.
#
# Наприклад, якщо слова:
# {"word": "cat"}
# {"word": "dog"}
# {"word": "runs"}
#
# DictVectorizer створить приблизно такі ознаки:
#
# word=cat   word=dog   word=runs
#     1          0          0
#     0          1          0
#     0          0          1
#
# fit_transform() одночасно:
# 1. знаходить усі можливі слова;
# 2. створює для них числові ознаки;
# 3. перетворює X_raw у матрицю X_vec.
X_vec = vec.fit_transform(X_raw)


# Розділяємо весь набір даних на навчальну та тестову частини.
#
# X_vec — ознаки всіх слів.
# y — правильні POS-теги всіх слів.
#
# test_size=0.2 означає:
# 80% даних → навчання моделі
# 20% даних → перевірка моделі
#
# random_state=42 фіксує випадковість,
# щоб при повторному запуску отримувати такий самий поділ.
X_train, X_test, y_train, y_test = train_test_split(
    X_vec,
    y,
    test_size=0.2,
    random_state=42
)


# Створюємо модель логістичної регресії.
#
# LogisticRegression використовується для класифікації.
#
# max_iter=1000 дозволяє алгоритму виконати максимум
# 1000 ітерацій під час навчання.
#
# Це збільшує шанс, що модель встигне знайти оптимальні параметри.
clf = LogisticRegression(max_iter=1000)


# Навчаємо модель на навчальній вибірці.
#
# X_train — ознаки слів.
# y_train — правильні POS-теги.
#
# Модель намагається навчитися залежності:
#
# слово → POS-тег
#
# Наприклад:
# "dog" → NN
# "run" → VB
# "the" → DT
clf.fit(X_train, y_train)


# Використовуємо навчену модель для прогнозування тегів
# для тестової вибірки.
#
# X_test містить слова, які модель не використовувала
# під час навчання.
#
# y_pred — передбачені моделлю POS-теги.
y_pred = clf.predict(X_test)


# Виводимо детальний звіт про якість роботи моделі.
#
# y_test — правильні теги.
# y_pred — теги, передбачені моделлю.
#
# classification_report показує:
#
# precision — наскільки точними були передбачення класу;
# recall — яку частину правильних об'єктів модель знайшла;
# f1-score — середнє між precision та recall;
# support — кількість прикладів кожного класу.
print(classification_report(y_test, y_pred))
```
